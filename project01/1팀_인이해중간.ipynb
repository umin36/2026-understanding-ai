```python
essentials = data[['fecha_dato', 'ncodpers', 'ind_empleado', 'pais_residencia', 'sexo', 'age', 'tiprel_1mes', 'nomprov', 'renta', 'segmento'] + goods]
```

```python
# Q1
q1_data = essentials.loc[essentials['fecha_dato']=='2016-05-28'].loc[:, ['nomprov', 'ncodpers', 'renta', 'age']]
q1_data
```

```python
q1_data = q1_data.dropna()
```

```python
grouped = q1_data.groupby('nomprov', as_index=True)

customer_num = grouped[['ncodpers',]].count().rename(columns={'ncodpers':'고객 수'})
income_mean = grouped[['renta',]].mean().rename(columns={'renta':'평균 소득'})
income_median = grouped[['renta',]].median().rename(columns={'renta':'소득 중앙값'})
age_mean = grouped[['age',]].mean().rename(columns={'age':'평균 나이'})
```

```python
features = customer_num.join(income_mean, on='nomprov').join(income_median, on='nomprov').join(age_mean, on='nomprov')
features
```

```python
features = features[features['고객 수'] >= 50]

# 평균 소득 순으로 내림차순 정렬 후 앞 5개 출력
print(list(features.sort_values(by='평균 소득', ascending=False).iloc[:5].index))
```

```python
# Q3
q3_data = essentials.loc[essentials['fecha_dato']=='2016-05-28'].loc[:, ['sexo', 'age', 'segmento', 'ind_empleado', 'tiprel_1mes']+goods]
q3_data
```

```python
q3_data = q3_data.dropna()
```

```python
X = q3_data[['sexo', 'age', 'segmento', 'ind_empleado', 'tiprel_1mes']].copy()
def _age_to_categorical(age):
    if age < 30:
        return 'young'
    if age < 65:
        return 'adult'
    return 'senior'
X['age'] = X['age'].apply(_age_to_categorical)
X
```

```python
y = q3_data[goods].sum(axis=1)
y = y>=3
y
```

```python
def observed(feature, target):
    df = pd.DataFrame(feature.values, columns=['feature']).join(pd.DataFrame(target.values, columns=['target']))
    feature_list = list(set(feature))
    target_list = list(set(target))
    observed = [[((df['target']==t) & (df['feature']==f)).sum() for f in feature_list] for t in target_list]
    observed = pd.DataFrame(observed, columns=feature_list, index=target_list)
    return observed

observed(X['age'], y)
```

```python
def expected(feature, target):
    feature_list = list(set(feature))
    target_list = list(set(target))
    target_ratio = np.array([(target==t).sum() for t in target_list]) / target.count()
    expected = [[r * (feature==f).sum() for f in feature_list] for r in target_ratio]
    expected = pd.DataFrame(expected, columns=feature_list, index=target_list)
    return expected

expected(X['age'], y)
```

```python
def chi2(feature, target):
    ob = observed(feature, target)
    ex = expected(feature, target)
    return (ob - ex)**2 / ex

chi2(X['age'], y)
```

```python
cols = X.columns
chi2_test = [chi2(X[col], y).values.sum() for col in cols]
chi2_test = pd.Series(chi2_test, index=cols)
chi2_test
```

```python
print(chi2_test.sort_values(ascending=False).index[0])
```
